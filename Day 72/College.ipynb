```python
import pandas as pd

df = pd.read_csv("salaries.csv")
print(df.columns)
```

```python
clean_df = df.dropna()
clean_df.head()
```

```python
# What college major has the highest mid-career salary? How much do graduates with this major earn?
highest_median = clean_df["Mid-Career Median Salary"].max()
row = clean_df["Mid-Career Median Salary"].idxmax()
major_highest_salary = clean_df["Undergraduate Major"].loc[row]
print(f'College major with highest mid-career salary is {major_highest_salary}.')
print(f'Graduates with this major earn {highest_median}.')
```
