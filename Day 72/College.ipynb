```python
import pandas as pd

df = pd.read_csv("salaries.csv")
print(df.columns)
```

```python
clean_df = df.dropna()
clean_df.head()
```

```python
# What college major has the highest mid-career salary? How much do graduates with this major earn?
highest_median = clean_df["Mid-Career Median Salary"].max()
row = clean_df["Mid-Career Median Salary"].idxmax()
major_highest_salary = clean_df["Undergraduate Major"].loc[row]
print(f'College major with highest mid-career salary is {major_highest_salary}.')
print(f'Graduates with this major earn {highest_median}.')
```

```python
# Which college major has the lowest starting salary and how much do graduates earn after university?
lowest_s_median = clean_df["Starting Median Salary"].min()
row = clean_df["Starting Median Salary"].idxmin()
major_lowest_salary = clean_df["Undergraduate Major"].loc[row]
print(f'College major with lowest median starting salary is {major_lowest_salary}.')
print(f'Graduates with this major earn {lowest_s_median}.')
```

```python
# Which college major has the lowest mid-career salary and how much can people expect to earn with this degree? 
lowest_m_salary = clean_df["Mid-Career Median Salary"].min()
row = clean_df["Mid-Career Median Salary"].idxmin()
major_m_lowest = clean_df["Undergraduate Major"].loc[row]
print(f'College major with lowest median mid-career salary is {major_m_lowest}.')
print(f'Graduates with this major earn {lowest_m_salary}.')
```

```python
# Using the .sort_values() method, can you find the degrees with the highest potential? Find the top 5 degrees with the highest values in the 90th percentile. 

sorted_df = clean_df.sort_values(by="Mid-Career 90th Percentile Salary", ascending=False)
sorted_df["Undergraduate Major"].head()
```

```python
# Also, find the degrees with the greatest spread in salaries. Which majors have the largest difference between high and low earners after graduation.

spread_col = clean_df["Mid-Career Median Salary"] - clean_df["Starting Median Salary"]
clean_df.insert(1, "Spread", spread_col)
sort_df = clean_df.sort_values(by="Spread", ascending=False)
sort_df["Undergraduate Major"].head()
```
